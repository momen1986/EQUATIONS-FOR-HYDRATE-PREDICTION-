```python
# ============================================================
# EQL EXTERNAL VALIDATION
# RRMSE (%) + ABSOLUTE ERROR (K) vs PRESSURE
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. DATA
# ------------------------------------------------------------

data = {
    "Sample": (
        ["Levan_3"] * 12
        + ["Sample_11"] * 10
        + ["Salman"] * 13
    ),

    "Pressure_psia": [
        690, 790, 890, 990, 1090, 1190, 1290, 1390,
        1490, 1990, 2461.67, 2761.67,

        590, 690, 790, 890, 990, 1090, 1190, 1290,
        1440, 1600,

        700, 850, 1000, 1150, 1300, 1450, 1600, 1800,
        2000, 2250, 2500, 2750, 3000
    ],

    "HFT_Actual_K": [
        285.9, 286.74, 287.52, 288.19, 288.77, 289.38,
        290.05, 290.69, 291.32, 292.51, 293.42, 294.07,

        287.94, 288.61, 289.18, 289.73, 290.31, 290.88,
        291.43, 292.04, 293.12, 294.26,

        289.86, 290.62, 291.27, 291.91, 292.48, 293.05,
        293.61, 294.32, 295.01, 295.76, 296.39, 296.94,
        297.49
    ],

    "HFT_Predicted_K": [
        285.3793547998587, 286.65737060517563, 287.8185787198052,
        288.78342150291536, 289.5161262552184, 290.02774000879407,
        290.37026320748004, 290.6220895414522, 290.86634137512016,
        292.8980162653478, 293.536220687614, 294.1334589028072,

        285.4811885161352, 286.8566617679097, 288.2373792123507,
        289.4882835853499, 290.51667702117516, 291.2837250954223,
        291.8040156436531, 292.1372072300574, 292.4803791339845,
        292.92389314020323,

        287.16212963156073, 289.2071729456165, 290.82107316719095,
        291.84443974189725, 292.38228179173467, 292.71522630889956,
        293.1317185514253, 294.0009367083488, 294.94084492220674,
        295.5448975291044, 295.6676305151474, 296.2776689795067,
        297.65676481965625
    ]
}

df = pd.DataFrame(data)

# ------------------------------------------------------------
# 2. ERROR CALCULATIONS
# ------------------------------------------------------------

df["Error_K"] = (
    df["HFT_Predicted_K"] - df["HFT_Actual_K"]
)

df["Abs_Error_K"] = df["Error_K"].abs()

# ------------------------------------------------------------
# 3. RRMSE CALCULATION
#
# RRMSE (%) = RMSE / mean(actual) * 100
# ------------------------------------------------------------

def calculate_rrmse(group):

    rmse = np.sqrt(
        np.mean(
            (group["HFT_Predicted_K"] -
             group["HFT_Actual_K"]) ** 2
        )
    )

    rrmse = (
        rmse /
        group["HFT_Actual_K"].mean()
        * 100
    )

    return rrmse


rrmse_table = (
    df.groupby("Sample")
      .apply(calculate_rrmse)
      .reset_index(name="RRMSE_%")
)

# Overall RRMSE
overall_rmse = np.sqrt(
    np.mean(
        (df["HFT_Predicted_K"] -
         df["HFT_Actual_K"]) ** 2
    )
)

overall_rrmse = (
    overall_rmse /
    df["HFT_Actual_K"].mean()
    * 100
)

print("\nRRMSE RESULTS")
print("=" * 50)
print(rrmse_table.to_string(index=False))

print("\nOverall RRMSE = {:.4f}%".format(overall_rrmse))


# ============================================================
# 4. FIGURE 1 — RRMSE BY EXTERNAL SAMPLE
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    rrmse_table["Sample"],
    rrmse_table["RRMSE_%"]
)

plt.ylabel("RRMSE (%)")
plt.xlabel("External Validation Sample")
plt.title("EQL External Validation: RRMSE")

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()

# Save high-resolution figure
plt.savefig(
    "EQL_External_Validation_RRMSE_1080dpi.png",
    dpi=1080,
    bbox_inches="tight"
)

plt.savefig(
    "EQL_External_Validation_RRMSE_1080dpi.tiff",
    dpi=1080,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 5. FIGURE 2 — ABSOLUTE ERROR vs PRESSURE
# ============================================================

plt.figure(figsize=(9, 6))

for sample, group in df.groupby("Sample"):

    plt.plot(
        group["Pressure_psia"],
        group["Abs_Error_K"],
        marker="o",
        linewidth=1.8,
        label=sample
    )

plt.xlabel("Pressure (psia)")
plt.ylabel("Absolute Error (K)")
plt.title("EQL External Validation: Absolute Error vs Pressure")

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.legend()

plt.tight_layout()

# Save high-resolution figure
plt.savefig(
    "EQL_External_Validation_AbsError_1080dpi.png",
    dpi=1080,
    bbox_inches="tight"
)

plt.savefig(
    "EQL_External_Validation_AbsError_1080dpi.tiff",
    dpi=1080,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 5. FIGURE 2 — ABSOLUTE ERROR vs PRESSURE
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

for sample, group in df.groupby("Sample"):

    plt.plot(
        group["Pressure_psia"],
        group["Abs_Error_K"],
        marker="o",
        linewidth=1.8,
        label=sample
    )

plt.xlabel("Pressure (psia)")
plt.ylabel("Absolute Error (K)")
plt.title("EQL External Validation: Absolute Error vs Pressure")

plt.grid(
    True,
    linestyle="--",
    alpha=0.5
)

plt.legend()
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. OPTIONAL — ERROR SUMMARY
# ------------------------------------------------------------

summary = (
    df.groupby("Sample")
      .agg(
          RMSE_K=(
              "Error_K",
              lambda x: np.sqrt(np.mean(x**2))
          ),
          MAE_K=(
              "Abs_Error_K",
              "mean"
          ),
          Max_Abs_Error_K=(
              "Abs_Error_K",
              "max"
          )
      )
      .reset_index()
)

summary["RRMSE_%"] = (
    summary["RMSE_K"]
    /
    df.groupby("Sample")["HFT_Actual_K"].mean().values
    * 100
)

print("\nEXTERNAL VALIDATION SUMMARY")
print("=" * 70)
print(summary.to_string(index=False))
```

```python

# ============================================================
# ONE BAR CHART: MEAN RRMSE BY PRESSURE RANGE
# Average across all methods and all three external samples
# NaN values are excluded
# Sample standard deviation (SD)
# Export resolution: 1080 DPI
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# SETTINGS
# ============================================================

pressure_ranges = [
    "Low\n(P <= 1000 psia)",
    "Medium\n(1000 < P <= 2000 psia)",
    "High\n(P > 2000 psia)"
]

summary_labels = [
    "Low (<=1000 psia)",
    "Medium (1000-2000 psia)",
    "High (>2000 psia)"
]

samples = ["Levan_3", "Sample_11", "Salman"]

output_folder = Path("Pressure_Range_RRMSE_Figures")
output_folder.mkdir(parents=True, exist_ok=True)

mean_rrmse = []
std_rrmse = []
sample_counts = []

# ============================================================
# VALIDATE INPUT DATA
# ============================================================

if "results" not in globals():
    raise NameError(
        "The dictionary 'results' was not found. "
        "Define the results dictionary before running this code."
    )

for sample in samples:
    if sample not in results:
        raise KeyError(
            f"Missing sample '{sample}' in the results dictionary."
        )

    for model, values in results[sample].items():
        if len(values) != len(pressure_ranges):
            raise ValueError(
                f"{sample} - {model}: expected "
                f"{len(pressure_ranges)} pressure-range values, "
                f"but received {len(values)}."
            )

# ============================================================
# CALCULATE MEAN RRMSE AND SAMPLE STANDARD DEVIATION
# ============================================================

for i, pressure_range in enumerate(pressure_ranges):

    values = []

    for sample in samples:

        for model, model_values in results[sample].items():

            value = model_values[i]

            # Exclude NaN and infinite values
            if np.isfinite(value):
                values.append(float(value))

    values = np.asarray(values, dtype=float)

    if len(values) == 0:
        mean_value = np.nan
        std_value = np.nan

    else:
        mean_value = np.mean(values)

        # Sample SD (ddof=1)
        std_value = (
            np.std(values, ddof=1)
            if len(values) > 1
            else 0.0
        )

    mean_rrmse.append(mean_value)
    std_rrmse.append(std_value)
    sample_counts.append(len(values))

    # Print numerical results
    if len(values) > 0:
        print(
            f"{pressure_range.replace(chr(10), ' ')}: "
            f"Mean RRMSE = {mean_value:.4f}%, "
            f"Sample SD = {std_value:.4f}%, "
            f"N = {len(values)}"
        )
    else:
        print(
            f"{pressure_range.replace(chr(10), ' ')}: "
            "No available values."
        )

# Convert results to arrays
plot_means = np.asarray(mean_rrmse, dtype=float)
plot_stds = np.asarray(std_rrmse, dtype=float)

valid = np.isfinite(plot_means) & np.isfinite(plot_stds)

# ============================================================
# PLOT
# ============================================================

fig, ax = plt.subplots(figsize=(10, 6))

x = np.arange(len(pressure_ranges))

bars = ax.bar(
    x[valid],
    plot_means[valid],
    width=0.55,
    yerr=plot_stds[valid],
    capsize=6,
    edgecolor="black",
    linewidth=0.8,
    label="Mean RRMSE"
)

# ============================================================
# ADD MEAN LABELS AND NUMBER OF VALUES
# ============================================================

for bar, idx in zip(bars, np.where(valid)[0]):

    height = bar.get_height()
    error = plot_stds[idx]

    # Mean RRMSE label above the error bar
    ax.annotate(
        f"{height:.4f}%",
        xy=(
            bar.get_x() + bar.get_width() / 2,
            height + error
        ),
        xytext=(0, 7),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

    # Number of available model-sample values
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        height * 0.50,
        f"N = {sample_counts[idx]}",
        ha="center",
        va="center",
        fontsize=10
    )

# ============================================================
# AXES AND FORMATTING
# ============================================================

ax.set_xticks(x)
ax.set_xticklabels(pressure_ranges, fontsize=10)

ax.set_ylabel("Mean RRMSE (%)", fontsize=12)
ax.set_xlabel("Pressure Range", fontsize=12)

ax.set_title(
    "Mean RRMSE Across Methods and External Samples",
    fontsize=14,
    fontweight="bold"
)

# Set y-axis limit to provide space for error bars and labels
if np.any(valid):

    upper_limit = np.max(
        plot_means[valid] + plot_stds[valid]
    )

    ax.set_ylim(
        0,
        upper_limit * 1.30 if upper_limit > 0 else 1
    )

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)

ax.set_axisbelow(True)

plt.tight_layout()

# ============================================================
# SAVE FIGURE — 1080 DPI
# ============================================================

output_file = (
    output_folder /
    "Mean_RRMSE_All_Pressure_Ranges_1080dpi.png"
)

plt.savefig(
    output_file,
    dpi=1080,
    bbox_inches="tight",
    format="png"
)

plt.show()
plt.close(fig)

# ============================================================
# SUMMARY TABLE
# ============================================================

summary_df = pd.DataFrame({
    "Pressure Range": summary_labels,
    "Mean RRMSE (%)": mean_rrmse,
    "Sample SD (%)": std_rrmse,
    "N": sample_counts
})

print("\n" + "=" * 80)
print("SUMMARY: MEAN RRMSE BY PRESSURE RANGE")
print("=" * 80)

print(
    summary_df.to_string(
        index=False,
        na_rep="N/A",
        formatters={
            "Mean RRMSE (%)": lambda value: (
                f"{value:.4f}" if pd.notna(value) else "N/A"
            ),
            "Sample SD (%)": lambda value: (
                f"{value:.4f}" if pd.notna(value) else "N/A"
            )
        }
    )
)

print("\n" + "=" * 80)
print(f"Figure saved to: {output_file}")
print("=" * 80)
```

```python

# ============================================================
# ONE GROUPED BAR CHART
# Mean RRMSE for each method across external samples
# Low, Medium, and High pressure ranges
# NaN and infinite values are excluded from averaging
# Export resolution: 1080 DPI
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# SETTINGS
# ============================================================

pressure_labels = [
    "Low\n(P <= 1000 psia)",
    "Medium\n(1000 < P <= 2000 psia)",
    "High\n(P > 2000 psia)"
]

samples = ["Levan_3", "Sample_11", "Salman"]

output_folder = Path("Pressure_Range_RRMSE_Figures")
output_folder.mkdir(parents=True, exist_ok=True)

# Consistent method ordering across all samples
methods = list(results["Levan_3"].keys())

# Validate the required samples and methods
for sample in samples:

    if sample not in results:
        raise KeyError(f"Missing sample: {sample}")

    for method in methods:

        if method not in results[sample]:
            raise KeyError(
                f"Method '{method}' is missing from '{sample}'."
            )

        if len(results[sample][method]) != 3:
            raise ValueError(
                f"{sample} - {method} must contain exactly "
                "three pressure-range values."
            )

# ============================================================
# CALCULATE MEAN RRMSE FOR EACH METHOD AND PRESSURE RANGE
# ============================================================

mean_results = {}
count_results = {}

for method in methods:

    mean_results[method] = []
    count_results[method] = []

    for pressure_index in range(3):

        values = []

        for sample in samples:

            value = results[sample][method][pressure_index]

            if np.isfinite(value):
                values.append(float(value))

        if values:
            mean_value = float(np.mean(values))
        else:
            mean_value = np.nan

        mean_results[method].append(mean_value)
        count_results[method].append(len(values))

# ============================================================
# PRINT SUMMARY TABLE
# ============================================================

rows = []

for method in methods:

    vals = mean_results[method]
    counts = count_results[method]

    rows.append({
        "Method": method,
        "Low RRMSE (%)": vals[0],
        "Medium RRMSE (%)": vals[1],
        "High RRMSE (%)": vals[2],
        "Low N": counts[0],
        "Medium N": counts[1],
        "High N": counts[2]
    })

summary_df = pd.DataFrame(rows)

print("\nMEAN RRMSE (%) BY METHOD AND PRESSURE RANGE")
print("=" * 110)

print(
    summary_df.to_string(
        index=False,
        na_rep="N/A",
        formatters={
            "Low RRMSE (%)": lambda v: (
                f"{v:.4f}" if pd.notna(v) else "N/A"
            ),
            "Medium RRMSE (%)": lambda v: (
                f"{v:.4f}" if pd.notna(v) else "N/A"
            ),
            "High RRMSE (%)": lambda v: (
                f"{v:.4f}" if pd.notna(v) else "N/A"
            )
        }
    )
)

# ============================================================
# PREPARE PLOT
# ============================================================

fig, ax = plt.subplots(figsize=(17, 8))

x = np.arange(len(methods))
bar_width = 0.25

low_values = np.array(
    [mean_results[m][0] for m in methods],
    dtype=float
)

medium_values = np.array(
    [mean_results[m][1] for m in methods],
    dtype=float
)

high_values = np.array(
    [mean_results[m][2] for m in methods],
    dtype=float
)

# ============================================================
# GROUPED BARS
# ============================================================

bars_low = ax.bar(
    x - bar_width,
    np.nan_to_num(low_values, nan=0.0),
    bar_width,
    label="Low pressure",
    edgecolor="black",
    linewidth=0.6
)

bars_medium = ax.bar(
    x,
    np.nan_to_num(medium_values, nan=0.0),
    bar_width,
    label="Medium pressure",
    edgecolor="black",
    linewidth=0.6
)

bars_high = ax.bar(
    x + bar_width,
    np.nan_to_num(high_values, nan=0.0),
    bar_width,
    label="High pressure",
    edgecolor="black",
    linewidth=0.6
)

# ============================================================
# NUMERICAL LABELS
# Missing values are not labelled
# ============================================================

bar_groups = [
    (bars_low, low_values),
    (bars_medium, medium_values),
    (bars_high, high_values)
]

for bars, original_values in bar_groups:

    labels = []

    for value in original_values:

        if np.isfinite(value):
            labels.append(f"{value:.4f}")
        else:
            labels.append("")

    ax.bar_label(
        bars,
        labels=labels,
        fontsize=8,
        padding=3,
        rotation=90
    )

# ============================================================
# AXES AND TITLE
# ============================================================

ax.set_xticks(x)

ax.set_xticklabels(
    methods,
    rotation=25,
    ha="right",
    fontsize=10
)

ax.set_ylabel(
    "Mean RRMSE (%)",
    fontsize=12
)

ax.set_xlabel(
    "Prediction Method",
    fontsize=12
)

ax.set_title(
    "Mean RRMSE by Prediction Method and Pressure Range",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.legend(
    title="Pressure Range",
    fontsize=10
)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.35
)

ax.set_axisbelow(True)

# ============================================================
# Y-AXIS LIMIT
# ============================================================

all_values = np.concatenate([
    low_values,
    medium_values,
    high_values
])

valid_values = all_values[np.isfinite(all_values)]

if len(valid_values) > 0:

    ymax = np.max(valid_values)

    ax.set_ylim(
        0,
        ymax * 1.25 if ymax > 0 else 1.0
    )

plt.tight_layout()

# ============================================================
# SAVE FIGURE — 1080 DPI
# ============================================================

output_file = (
    output_folder /
    "Mean_RRMSE_by_Method_and_Pressure_1080dpi.png"
)

plt.savefig(
    output_file,
    dpi=1080,
    bbox_inches="tight",
    format="png"
)

plt.show()
plt.close(fig)

# ============================================================
# PRINT OUTPUT LOCATION
# ============================================================

print("\n" + "=" * 75)
print("FIGURE SAVED SUCCESSFULLY")
print("=" * 75)

print(f"Output file: {output_file.resolve()}")
print("Resolution: 1080 DPI")
```

```python

import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# DATA
# Mean RRMSE and Population SD across the three external datasets
# ============================================================

models = [
    "PySR-SP_GR",
    "PySR-Composition",
    "XGB-Composition",
    "EQL-Composition",
    "Motiee (1991)",
    "XGB-SP_GR",
    "EQL-SP_GR"
]

mean_rrmse = np.array([
    0.2250,       # PySR-SP_GR
    0.250931,     # PySR-Composition (updated E6)
    0.2638,       # XGB-Composition (updated)
    0.4038,       # EQL-Composition
    0.2683,       # Motiee (1991)
    0.3523,       # XGB-SP_GR (updated)
    0.3658        # EQL-SP_GR (updated)
])

sd_values = np.array([
    0.0981,       # PySR-SP_GR
    0.049969,     # PySR-Composition (updated E6)
    0.1014,       # XGB-Composition (updated)
    0.0891,       # EQL-Composition
    0.0026,       # Motiee (1991)
    0.0611,       # XGB-SP_GR (updated)
    0.1531        # EQL-SP_GR (updated)
])

# ============================================================
# EQL-COMPOSITION
# Existing pressure restriction is retained
# ============================================================

eql_comp_sample_rrmse = np.array([
    0.30044898,   # Levan_3
    0.39305061,   # Sample_11
    0.51781940    # Salman
])

eql_comp_mean = np.mean(eql_comp_sample_rrmse)
eql_comp_sd = np.std(eql_comp_sample_rrmse, ddof=0)

# ============================================================
# EQL-SP_GR
# All available pressure points are included
# ============================================================

eql_spgr_sample_rrmse = np.array([
    0.149597,     # Levan_3
    0.483000,     # Sample_11
    0.464820      # Salman
])

eql_spgr_mean = np.mean(eql_spgr_sample_rrmse)
eql_spgr_sd = np.std(eql_spgr_sample_rrmse, ddof=0)

# Update EQL metrics using calculated values
eql_comp_idx = models.index("EQL-Composition")
mean_rrmse[eql_comp_idx] = eql_comp_mean
sd_values[eql_comp_idx] = eql_comp_sd

eql_spgr_idx = models.index("EQL-SP_GR")
mean_rrmse[eql_spgr_idx] = eql_spgr_mean
sd_values[eql_spgr_idx] = eql_spgr_sd

# ============================================================
# PRINT UPDATED EQL METRICS
# ============================================================

print("=" * 58)
print("EQL-COMPOSITION METRICS")
print("=" * 58)
print(f"Levan_3 RRMSE:   {eql_comp_sample_rrmse[0]:.6f}%")
print(f"Sample_11 RRMSE: {eql_comp_sample_rrmse[1]:.6f}%")
print(f"Salman RRMSE:    {eql_comp_sample_rrmse[2]:.6f}%")
print(f"Mean RRMSE:      {eql_comp_mean:.6f}%")
print(f"Population SD:   {eql_comp_sd:.6f}%")

print("\n" + "=" * 58)
print("EQL-SP_GR METRICS (ALL PRESSURE POINTS)")
print("=" * 58)
print(f"Levan_3 RRMSE:   {eql_spgr_sample_rrmse[0]:.6f}%")
print(f"Sample_11 RRMSE: {eql_spgr_sample_rrmse[1]:.6f}%")
print(f"Salman RRMSE:    {eql_spgr_sample_rrmse[2]:.6f}%")
print(f"Mean RRMSE:      {eql_spgr_mean:.6f}%")
print(f"Population SD:   {eql_spgr_sd:.6f}%")

# ============================================================
# REFERENCE MODELS
# ============================================================

reference_models = [
    "XGB-Composition",
    "Motiee (1991)",
    "XGB-SP_GR"
]

reference_colors = {
    "XGB-Composition": "darkorange",
    "Motiee (1991)": "red",
    "XGB-SP_GR": "purple"
}

# Remove reference models from the bars
plot_models = [
    m for m in models if m not in reference_models
]

plot_mean = np.array([
    mean_rrmse[models.index(m)] for m in plot_models
])

plot_sd = np.array([
    sd_values[models.index(m)] for m in plot_models
])

# ============================================================
# DOUBLE BAR CHART
# ============================================================

fig, ax = plt.subplots(figsize=(14, 7))

x = np.arange(len(plot_models))
width = 0.36

bars_mean = ax.bar(
    x - width / 2,
    plot_mean,
    width,
    label="Mean RRMSE",
    color="steelblue",
    edgecolor="black",
    zorder=2
)

bars_sd = ax.bar(
    x + width / 2,
    plot_sd,
    width,
    label="Population SD",
    color="mediumseagreen",
    edgecolor="black",
    zorder=2
)

# ============================================================
# REFERENCE LINES
# Dashed = Mean RRMSE
# Dotted = Population SD
# ============================================================

for model in reference_models:
    idx = models.index(model)

    ref_mean = mean_rrmse[idx]
    ref_sd = sd_values[idx]
    color = reference_colors[model]

    ax.axhline(
        y=ref_mean,
        color=color,
        linestyle="--",
        linewidth=2,
        label=f"{model} Mean = {ref_mean:.4f}%",
        zorder=1
    )

    ax.axhline(
        y=ref_sd,
        color=color,
        linestyle=":",
        linewidth=2,
        label=f"{model} SD = {ref_sd:.4f}%",
        zorder=1
    )

# ============================================================
# VALUE LABELS
# ============================================================

for bar in bars_mean:
    h = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        h + 0.006,
        f"{h:.4f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

for bar in bars_sd:
    h = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        h + 0.006,
        f"{h:.4f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

# ============================================================
# FORMATTING
# ============================================================

ax.set_xticks(x)
ax.set_xticklabels(plot_models, rotation=20, ha="right")

ax.set_ylabel("Mean RRMSE / Population SD (%)")
ax.set_xlabel("Model")

ax.set_title(
    "External Validation: Mean RRMSE and Population SD"
)

ax.grid(
    axis="y",
    linestyle=":",
    alpha=0.3,
    zorder=0
)

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.02, 1),
    fontsize=8,
    ncol=1
)

ax.set_ylim(0, 0.48)

plt.tight_layout()

plt.savefig(
    "Model_Stability_1080dpi.png",
    dpi=1080,
    bbox_inches="tight",
    format="png"
)

plt.show()
```
